### 2.3 Examine airline on-time reporting data.

In the upcoming steps, you will join the weather observation data with airline on-time reporting data.

Review the following query and resulting DataFrame, which reports a count of flight cancellations by date for flights originating from the San Francisco Airport during the calendar year 2019.

*You needn't edit anything in the following cell. Just run it.*

````python
# In the following steps, you will join the weather observation data with airline on-time reporting data.
# Review the resulting DataFrame to peek at the data you will be joining to.
sfo_2019_flight_df = (session
    .table("SNOWBEARAIR_DB.RAW.ONTIME_REPORTING")
    .select(
         col("FL_DATE")
        ,col("ORIGIN")
        ,col("CANCELLED")
      )
    .filter(col("CANCELLED") == lit(1.00))
    .filter((col("FL_DATE") >= to_date(lit("2019-01-01"))) & (col("FL_DATE") <= to_date(lit("2019-12-31"))))
    .filter(col("ORIGIN") == lit("SFO"))
    .groupBy(["FL_DATE","ORIGIN"])
    .agg([count(col("CANCELLED")).alias("CANCELLED_FLIGHT_COUNT")])
    .sort(col("CANCELLED_FLIGHT_COUNT").desc()))

sfo_2019_flight_df.show()
````

### 2.4 Join the flight cancellation data and weather data to examine trends.

To join the flight data with the weather data, match the following values:
- Match flight date from flight data with observation date from weather data
- Match origin airport code (IATA code) from flight data with station ICAO airport code from weather data

One challenge with this join is that the two data sets utilize different airport code standards. The weather data set uses ICAO airport codes and the flight data set uses IATA airport codes. While this wouldn’t work for every airport code conversion, the U.S. (contiguous states) airport ICAO codes match the IATA codes with an additional prefix of "K". 

For this exercise, convert the ICAO code from the weather data to an IATA code by taking a substring of the last three characters of the ICAO code.

> **&#128221; Note:** [Source: List of airports by ICAO code: K](https://en.wikipedia.org/wiki/List_of_airports_by_ICAO_code:_K)

#### 2.4.1 Convert the airport station code from the weather observations data set from an ICAO code to an IATA code.

Modify the previous set of transformations on the weather data set in order to convert the ICAO code to an IATA code.

To perform this conversion, simply extract the last three characters from the ICAO code using the `substring()` method, which accepts the following parameters: *column*, *starting position (one-based index)*, and *length of substring*.

*You needn't edit anything in the following cell. Just run it.*

````python
# Modify the previous set of transformations on the weather data in order to convert the ICAO code to an IATA code

sfo_2019_weather_df = (weather_2019_table_df
     .join_table_function(    
         "flatten"                # lateral flatten is assumed
        ,col("v")                 # Column to flatten
        ,lit("data:observations") # Path to element to flatten
        ,lit(False)               # Ignore null rows
        ,lit(False)               # Recursive?
        ,lit("ARRAY")             # Type to flatten
      )
     .select(       
         col("t").alias("observation_date")
        ,substring(
                (col("v")["station"]["ICAO"]).cast(StringType()),  # Drill down to ICAO 
                 lit(2),                                           # Extract substring starting at position 2
                 lit(3)                                            # Grab 3 chars
                ).alias("iata")                                    # Rename column to IATA, used for grouping later
        ,(col("value")["air"]["temp"]).cast(FloatType()).alias("air_temp")
        ,(col("value")["air"]["dew-point"]).cast(FloatType()).alias("dew_point")
        ,(col("value")["visibility"]["distance"]).cast(IntegerType()).alias("visibility")
      )
     .filter(col("IATA") == lit("SFO"))
     .filter(col("air_temp") != lit(999.9))
     .group_by(
         ["iata","observation_date"]
      )
     .agg(
         [
           min(col("air_temp")).alias("min_air_temp")
          ,max(col("air_temp")).alias("max_air_temp")
          ,round(avg(col("air_temp")),lit(2)).alias("avg_air_temp")
          ,round(avg(col("dew_point")),lit(2)).alias("avg_dew_point")
          ,round(avg(col("visibility")),lit(2)).alias("avg_visibility")
         ]
      ))

sfo_2019_weather_df.show()
````
