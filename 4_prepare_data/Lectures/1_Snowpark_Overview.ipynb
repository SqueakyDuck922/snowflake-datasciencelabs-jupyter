## 1. What Is Snowpark?

So what is Snowpark?

- Software Development Kit (**SDK**)
    - Application programming interface (**API**)
    - User-Defined Functions (**UDFs**)
    - Stored Procedures (**SPs**)
- Interact programmatically with data
- Push down data processing to Snowflake Virtual Warehouse
    - Scalable compute
    - No application servers/clusters to maintain
    - Secure sandbox environment
- **Snowpark Language Support**: (GA for all cloud providers and editions)
    - **Scala**
    - **Java** 
    - **Python**

> In this course, we will focus primarily on Python.

### 5A. What Is a DataFrame?

- Key abstraction of Snowpark API
- Tabular data representation
    - One or more columns
    - Zero or more rows
- Can represent client-side data; or server-side table, view, stage file, or query results
- Each column has a data type
- Lazily executed
- Immutable
- Similar to other dataframe programming constructs (pandas, Spark)
- For full details of Snowpark-specific `DataFrame` capabilities, see: [snowflake.snowpark.DataFrame](https://docs.snowflake.com/en/developer-guide/snowpark/reference/python/api/snowflake.snowpark.DataFrame.html#snowflake-snowpark-dataframe)

<a id="dataframe_operations"></a>
### 5B. DataFrame Operations: Create, Transform, and Action

#### Create operations do not materialize data.
- Create operations define `DataFrame` objects
- The object is just a shell until an action is invoked
- Like choosing a recipe to bake and noting the ingredients required

#### Transformations do not materialize data.
- Executed lazily; no data is materialized until an action is invoked
- `DataFrames` are immutable and each transformation returns a new `DataFrame` object
- Like a recipe, the steps are noted

#### Actions materialize data.
- Executed eagerly
- Once an action is invoked, all previously noted create and transform operations for the `DataFrame` are evaluated
- Statements are executed on the server and can be seen in query history
- Actions let you see, gather, or persist `DataFrame` data
- Like gathering ingredients and executing the steps of a recipe and then... you get a cake!

### 6A. Connecting to a Snowflake Account

Connecting your Snowpark application to a Snowflake account is simple. The Snowpark API provides "builder" to facilitate configuring your connection and creating a `snowflake.snowpark.Session` object. 

#### Connection Properties

- Connection managed by Snowpark `Session` object
- Use the `SessionBuilder` class to create a `Session`
- Configure connection properties in a `Dict` object

Creating a `Session` using a `dict` of key-value pairs:

```python
my_properties_dictionary = ([
    "user": "<user_name>",
     "password": "<password>",
     "account": "<account_name>",
     "role": "<role_name>",
     "warehouse": "<warehouse_name>",
     "database": "<database_name>",
     "schema": "<schema1_name>"
 ]
)
my_session_builder = Session.builder
my_session_builder.configs(my_properties_dictionary)
session = my_session_builder.create()
```

Example using function chaining:

```python
session = (Session
    .builder                            # Returns a Session.SessionBuilder object
    .configs(my_properties_dictionary)  # Configures the SessionBuilder object
    .create()                           # Creates the configured Session object 
)
```

> **&#128221; Note:** See documentation for further details:
> - [snowflake.snowpark.Session.builder](https://docs.snowflake.com/en/developer-guide/snowpark/reference/python/api/snowflake.snowpark.Session.builder.html)

### 6B. Using Sessions

A Snowpark `Session` object is your "hook" into your Snowflake account and is central to the Snowpark API. Use your `Session` to:
- Create a `SessionBuilder` to connect to a Snowflake account
- Execute SQL statements

```python
    my_df1 = session.sql("select * from ....")

```
- Create, transform, and perform actions on `DataFrame`s

```python
    my_df2 = (session
        .read
        .option("...")
        .csv(pathToStagedFile)        # Created from a staged file
    )

    my_df3 = (session
        .table("...")                 # Created from a table or a view
        .where("...").limit("...")    # Transform
    )

    my_list_of_row_objects = (session
        .table("...")                 # Created from a table or a view
        .select("...").where("...").  # Transform
        .collect("...")               # Action, returns a list of Row objects
    )

```
- Reflect on context settings

```python
    db_name = session.get_current_database()

```

- Change context settings

```python
   session.use_database("My_DB")

```

- Reflect on platform settings

```python
   imports_list = session.get_imports()

```

To get started with a `Session` in your Snowpark application:
1. Import the Snowpark `Session` class
2. Create a `SessionBuilder` by calling `Session.builder`
3. Provide a `Dict` containing connection parameters
4. Invoke `create` on the `SessionBuilder` object to create the `Session`

> **&#128221; Note:** See documentation for further details:
> - [Executing SQL Statements](https://docs.snowflake.com/en/developer-guide/snowpark/python/working-with-dataframes.html#executing-sql-statements)
> - [snowflake.snowpark.Session.sql(...)](https://docs.snowflake.com/en/developer-guide/snowpark/reference/python/api/snowflake.snowpark.Session.sql.html#snowflake.snowpark.Session.sql)
> - [snowflake.snowpark.Session.use_database(..)](https://docs.snowflake.com/en/developer-guide/snowpark/reference/python/api/snowflake.snowpark.Session.use_database.html#snowflake.snowpark.Session.use_database)
> - [snowflake.snowpark.Session.get_current_database()](https://docs.snowflake.com/en/developer-guide/snowpark/reference/python/api/snowflake.snowpark.Session.get_current_database.html#snowflake.snowpark.Session.get_current_database)
> - [snowflake.snowpark.Session.get_imports()](https://docs.snowflake.com/en/developer-guide/snowpark/reference/python/api/snowflake.snowpark.Session.get_imports.html#snowflake.snowpark.Session.get_imports)


````python
# Import the Session class
from snowflake.snowpark import Session

# Declare a variable to hold the name of our properties file 
filename = "connect.properties"
print(f"The properties file with connection information is named {filename}")

# Create a dictionary to hold connection parameters
connection_parameters = {}

# Read connection properties into a dictionary
print("Reading key/pair combinations from that file into a Python dictionary")
with open(filename) as file:
    for line in file:
       (key, value) = line.split("=")
       connection_parameters[key.strip()] = value.strip()

# Create the Session using a SessionBuilder object
print("Using that dictionary with Session.builder.configs(....) to create a Session instance named 'session'")
session = (Session
    .builder                        # Creates a SessionBuilder instance
    .configs(connection_parameters) # Configures and returns the same SessionBuilder instance
    .create()                       # Creates and returns the Session object
)

print(f"Session created: {str(type(session))} ")
````

<a id="creating_snowflake_objects_for_this_class"></a>
#### Create Snowflake Objects for this Class

For this class, you will need a warehouse of your own as well as exclusive use of a database within the classroom Snowflake account. The Snowpark code below will create these objects. Don't be thrown by `import`, `.sql(...)`, `.collect()`, `.show(...)` and other API calls. We will cover **all** the features used below in subsequent lessons. 

For now, just look for the final message `========== READY ===========`. 

````python
# Create an empty DataFrame for reflecting upon our context items
context_df = session.create_dataframe([""]).to_df("")

# Hard code the lesson name
lesson_name = "SNOWPARK_OVERVIEW_PY"

# Retrieve the current username
from snowflake.snowpark.functions import current_user
username = (str(context_df
    .select(current_user())
    .collect()[0][0]
   )
)
print(f"The current user is: {username}")

# Dynamically name our demo objects
db_name = f"{username}_DB"
schema_name = f"{lesson_name}_LESSON"
full_schema_name = f"{db_name}.{schema_name}"
wh_name = f"{username}_WH"

# The status column will be used when confirming, creating, and setting the context items
from snowflake.snowpark.functions import col
status_column = col("\"status\"")

# Set our current role appropriately
role = "TRAINING_ROLE"
print(f"Telling the session to use the role {role}")
session.use_role(role)

# Set the query tag for operations in this lesson
tag = f"Snowpark Programming: {username} - {lesson_name}"
print(f"Setting the query_tag of the Session instance to: {tag}")
session.query_tag = tag

# Verify query tag and print it
print(f"Query tag is: {session.query_tag}")

# Ensure we have a warehouse set in our context with the proper configuration
print(f"Creating a warehouse named {wh_name} if it doesn't exist.")
(session.sql(f"CREATE WAREHOUSE IF NOT EXISTS {wh_name}")
    .rename(status_column, f"Created warehouse {wh_name}?")
    .show(1,80)
)

print(f"Altering the warehouse to set auto_resume to true and the size to XSMALL.")
(session.sql(f"ALTER WAREHOUSE IF EXISTS {wh_name} SET AUTO_RESUME = true warehouse_size = XSMALL")
    .rename(status_column, f"Altered warehouse {wh_name}?")
    .show(1,80)
)

print(f"Setting current warehouse for the Session instance to {wh_name}")
session.use_warehouse(wh_name)

# Create (if not present) and confirm the database and schema
print(f"Creating database {db_name}")
(session.sql(f"CREATE DATABASE IF NOT EXISTS {db_name}")
    .rename(status_column,f"Created database {db_name}?")
    .show(1,80)
) 

print(f"Creating schema {schema_name}")
(session.sql(f"CREATE OR REPLACE SCHEMA {full_schema_name}")
    .rename(status_column, f"Created schema {full_schema_name}?")
    .show(1,80)
) 

# Set our current namespace appropriately
print(f"Setting current namespace in the Session instance to {full_schema_name}")
session.use_schema(full_schema_name)

# Retrieve the account name
print("Retrieving the current account name from the Session instance")
acct_name = session.get_current_account()

# Confirm our context items
from snowflake.snowpark import functions
print("The current context items:")
(context_df
    .select(
         functions.current_database().alias("Database")
        ,functions.current_schema().alias("Schema")    
        ,functions.current_warehouse().alias("Warehouse")
        ,functions.current_role().alias("Role") 
        ,functions.current_user().alias("User") 
        ,functions.lit(acct_name).alias("Account") # As of 1.5.1, functions has no current_account() function
      )
    .show(1,80)
)
print("========== READY ===========")
````

### 6C. View Snowpark-Generated SQL in Query History in Snowsight

As mentioned above, when an action is encountered in Snowpark code, the logic behind the `DataFrame` is converted to SQL and sent to the Snowflake account that is configured in your `Session` object. 

The following Snowpark code creates a `DataFrame`, performs some basic transformations, and then saves the result to a Snowflake table. Once the action `.save_as_table()` is invoked, you can view the SQL that was sent to the Snowflake using the Query History of the Snowsight interface. (More details regarding writing to tables with the `.save_as_table()` will be covered later in the course.)

````python
from snowflake.snowpark.functions import col

(session.table("SNOWFLAKE_SAMPLE_DATA.TPCH_SF1.NATION")
     .select("N_NATIONKEY","N_NAME")        # A simple transformation
     .where(col("N_NAME").like("U%"))       # A filter transformation
     .order_by(col("N_NATIONKEY").asc())    # A sort transformation
     .write                                 # Covered in the Actions lession
     .save_as_table(table_name = "JUST_THE_U_NATIONS",mode="OVERWRITE")  # An action
)
````

## 7. Clean Up and Close Session

Best practice is to clean up demo objects, suspend the warehouse, and close the Snowpark session.

````python
from snowflake.snowpark.exceptions import SnowparkSQLException, SnowparkSessionException
session_exists = True

# Drop the schema created at the beginning of our session
try:
    (session.sql(f"DROP SCHEMA IF EXISTS {full_schema_name}")
        .show(1,80)
    )
except SnowparkSessionException as ex:    
    print(f"Trouble dropping schema {full_schema_name}.")
    print(f"Message: {str(ex)}")
    session_exists = False
````

````python
# First try to suspend the warehouse
try:
    if(session_exists):
        (session.sql(f"ALTER WAREHOUSE {wh_name} SUSPEND")
            .show(1,80)
        )
        print(f"Warehouse {wh_name} suspended")
    else:
        print("No session to available")
except Exception as ex:    
    print(f"Trouble suspending warehouse {wh_name}. It was possibly not running.")
    print(f"Message: {str(ex)}")
````

````python
# Then close the session    
try:
    if(session_exists):
        session.close() # Close the session object
        print("Session closed")
    else:
        print("No session to close")
except Exception as ex:    
    print(f"Message: {str(ex)}")    
````
