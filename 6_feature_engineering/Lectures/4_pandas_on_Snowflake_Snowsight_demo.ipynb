# Git repository

You can integrate your remote Git repository with Snowflake so that files from the repository are synchronized to a special kind of stage called a repository stage. The repository stage acts as a local Git repository with a full clone of the remote repository, including branches, tags, and commits.


With a remote Git repository integrated with your Snowflake account, you synchronize files from the remote repository to a repository stage in Snowflake. To access a file in Snowflake, you refer to it on the repository stage.
![](https://docs.snowflake.com/en/_images/git-architecture.png)


### Snowflake repository stage

A repository stage is the Snowflake representation of a repository. Like a local Git repository, it is a full clone with all branches, tags, and commits from the remote repository.

After repository contents are on the repository stage, you can reference files there as you would any other file on a stage. Note that while you can execute GET commands against a repository stage, most other ordinary stage commands aren’t applicable to a repository stage.

### Supported platforms
Currently you can integrate Git repositories on the following Git platforms:

- GitHub
- GitLab
- BitBucket
- Azure DevOps
- AWS CodeCommit




# Part 2: Introduction to pandas on Snowflake
The pandas on Snowflake API allows you to run your pandas code directly on your data in Snowflake. Built to replicate the functionality of pandas - including its data isolation and consistency guarantees - the pandas on Snowflake API enables you to scale up your traditional pandas pipelines with just a few lines of change.

pandas on Snowflake is an extension of the Snowpark API that unlocks the power of Snowflake for pandas developers. You will be able to get the same pandas-native experience you know and love with Snowflake's performance, scale and governance.


## Existing pandas code can be used with ease in pandas on Snowflake 

Below is an example of vanilla pandas.

```
import pandas as pd

# Create a Snowflake session
from snowflake.snowpark.context import get_active_session
session = get_active_session()

# Read data from a Snowflake table into a Pandas DataFrame
df = session.table("SNOWFLAKE_SAMPLE_DATA.TPCH_SF1.CUSTOMER").select("C_CUSTKEY","C_ACCTBAL").limit(10)

pdf = df.to_pandas()

# Manipulate the DataFrame using Pandas
pdf["ACCOUNT_BALANCE_WITH_SALES_TAX"] = pdf["C_ACCTBAL"] * 1.06

from tabulate import tabulate
print(tabulate(pdf, headers = "keys", tablefmt = "fancy_grid"))
```

Below is the same example using pandas on Snowflake. One import was added, one import was modified, and the function call was modified.

```
# Import plugin
import snowflake.snowpark.modin.plugin  

# Modify the import 
import modin.pandas as pd

# Create a Snowflake session
from snowflake.snowpark.context import get_active_session
session = get_active_session()

# Read data from a Snowflake table into a Pandas DataFrame
df = session.table("SNOWFLAKE_SAMPLE_DATA.TPCH_SF1.CUSTOMER").select("C_CUSTKEY","C_ACCTBAL").limit(10)

pdf = df.to_snowpark_pandas() # From Snowpark DataFrame to pandas on Snowflake 

# Manipulate the DataFrame using Pandas
pdf["ACCOUNT_BALANCE_WITH_SALES_TAX"] = pdf["C_ACCTBAL"] * 1.06

from tabulate import tabulate
print(tabulate(pdf, headers = "keys", tablefmt = "fancy_grid"))
```

## 1. Why use pandas on Snowflake? 
Already for many years pandas is the go-to data processing library for millions of people worldwide, including countless Snowflake users. However, pandas was never built to handle data at the scale organizations are operating today. 

Running pandas code requires transferring and loading all of the data into a single in-memory process. It becomes unwieldy on moderate-to-large data sets and breaks down completely on data sets that grow beyond what a single node can handle. 

We know organizations work with this volume of data today, and pandas on Snowflake enables you to execute that same pandas code, but with all the pandas processing pushed down to run in a distributed fashion in Snowflake. Your data never leaves Snowflake, and your pandas workflows can process much more efficiently using the Snowflake elastic engine. This brings the power of Snowflake to pandas developers everywhere.

## 2. Getting Started - Reading Data from Snowflake

We will start with reading some data from Snowflake. Just like we did with the Snowpark lecture, we are going to read data from our sample data.

`SNOWFLAKE_SAMPLE_DATA.TPCH_SF10.LINEITEM`

This table meets the industry-standard TPC-DS and TPC-H benchmarks for a proper sample dataset.

Let's:
- read the data
- measure how long it takes us



````
from time import perf_counter
start = perf_counter()
df = pd.read_snowflake("SNOWFLAKE_SAMPLE_DATA.TPCH_SF10.LINEITEM")
end = perf_counter()
data_size = len(df)
print(f"Took {end - start} seconds to read a table with {data_size} rows into pandas on Snowflake!")
snow_time = end - start
````

Now we know how long it takes to read about 60 million rows into pandas on Snowflake. 

But what does that mean?
- We read data
- Data stays in Snowflake
- You get the same pandas-native experience you know

It would take A LOT longer if we would have read this data to our local machine.

Let's print the first couple of rows.

````
start = perf_counter()
result_5 = df.head(5)
end = perf_counter()
print(result_5)
print(f"It took {end - start} seconds to show the first 5 rows.")
````

Above we see the difference between pandas on Snowflake and the Snowpark dataframe API.

The latter has lazy evaluation and when executing an action, such as `.show()` it would start reading the data.

However pandas on Snowflake mimics the eager behavior of pandas while still running everything in Snowflake. So when we now want to see the data, the computation time will be extremely low.

Let's compare:


````
test_df = session.table("SNOWFLAKE_SAMPLE_DATA.TPCH_SF10.LINEITEM")
````

However, with just reading a tiny amount of data, Snowpark will read only 1 Micro Partition and will also be quite fast.

````
start = perf_counter()
result = test_df.limit(5)
result.show()
end = perf_counter()
print(result)
print(f"It took {end - start} seconds to show the first 5 rows.")
````

The pandas on Snowflake and DataFrame API are highly interoperable, so you can build a pipeline that leverages both APIs.

You can use the following operations to do conversions between Snowpark DataFrames and pandas on Snowflake DataFrames:

````
pandas_test_df = result.to_snowpark_pandas()
print('pandas_test_df is of class:',  type(pandas_test_df))
print(pandas_test_df)
print(' ')

snowpark_df = pd.to_snowpark(result_5, index = False)
print('snowpark_df is of class:',  type(snowpark_df))
snowpark_df
````

````
# Input: Snowpark pandas DataFrame
# Output: native pandas DataFrame
pandas_df = pandas_test_df.to_pandas()
print(type(pandas_df))

# Input: small Dataframe
# Output: Snowpark pandas DataFrame
columns = ['Student ID', 'Course ID', 'Marks']
data = [(103, 201, 67), (103, 203, 67), (103, 204, 89)]
snowpark_pandas_df = pd.DataFrame(data, columns=columns)
print(type(snowpark_pandas_df))


# Input: pandas Dataframee or Snowpark pandas dataframe
# Output: Snowflake Table
snowpark_df = session.write_pandas(snowpark_pandas_df, "write_pandas_table", auto_create_table=True, table_type="temp")
print(type(snowpark_df))
````

## 2. Filtering

Let's take a look at some common data transformations - starting with filtering.

We would like to look at the data where the instruction was to deliver in person.

We learned in a previous module about StreamLit. We can use that here to show some tables. 



````
start = perf_counter()
inperson_df = df[(df['L_SHIPINSTRUCT'] == 'DELIVER IN PERSON')]
repr(inperson_df)
end = perf_counter()
st.dataframe(inperson_df.head())
print(f"Filtering for delivering in person took {end - start} seconds in pandas on Snowflake")
````

Let's try an even more granular filter.

````
start = perf_counter()
filtered_df = df[((df['L_SHIPMODE'] == 'AIR') | (df['L_SHIPMODE'] == 'RAIL') | (df['L_SHIPMODE'] == 'SHIP')) & (df['L_SHIPINSTRUCT'] == 'DELIVER IN PERSON')]
repr(filtered_df)
end = perf_counter()
st.dataframe(filtered_df.head())
print(f"Filtering for items transported by Plane, Ship and train took {end - start} seconds in pandas on Snowflake")
````
